```python
marks = float(input('Enter student marks: '))
if marks > 100 or marks < 0:
    result = f"Invalid marks i.e. {marks}"
elif 80 <= marks < 101:
    result = 'A1'
elif 70 <= marks < 80:
    result = 'A'
elif 60 <= marks < 70:
    result = 'B'
elif 50 <= marks < 60:
    result = 'C'
elif 40 <= marks < 50:
    result = 'D'
else:
    result = 'Better Luck Next Time'

print(result)
```
