```python
# print those cities having legth in even number 
# e.g output Multan, Lahore
cities = ['Karachi', 'Quetta', 'Lahore', 'Rawalpindi', 'Multan']
for city in cities:
    if len(city) % 2 == 0:
        print(city)
```

```python
# print those cities having more 2 'a' in city 
# e.g output: Multan, Lahore
cities = ['Karachi', 'Quetta', 'Lahore', 'Rawalpindi', 'Multan']
for city in cities:
    if city.count('a') == 2:
        print(city)
```
