```python
colors = ['Red', 'Blue']
sizes = ['Small', 'Medium','Large']

for color in colors:
    for size in sizes:
        print(f'size : {size}, color: {color}')
```

```python
# print those city/cities having all characters unique (not repetitive)
cities = ['Karachi','Lahore','Quetta','Dadu','Murre']

for city in cities:
    charcount = True
    for char in city:
        if city.lower().count(char) >1:
            charcount = False
            break
    # if charcount == True
    if charcount:
        print(city)
```

```python
for city in cities:
    empty_set = set()
    for char in city.lower():
        empty_set.add(char)
    if len(empty_set) == len(city):
        print(city)
```

# While Loop

```python
flag = True
# while condition
while flag:
    country = input('Enter your country name: ').lower()
    country = country[:8]
    if country == 'pakistan':
        print('You are eligible for cnic')
        flag = False
    else :
        print(f'This country {country} is not eligible')
```

```python
length = float(input('Enter length'))
width = float(input('Enter width'))

if length == width:
    print(length * width)
else:
    print(2 * (length + width))
```
